Q13

```python
import pandas as pd
import numpy as np
df=pd.read_csv(student_performance)
df_subset=df.head(6).copy()
df_subset["Python_Score"]=[95, 70, 88, 55, 92, 78]
df_subset["SQL_Score"]=[90, 75, 85, 60, 94, 72]
df_subset["Pandas_Score"]=[88, 72, 91, 58, 89, 75]
df_subset["NumPy_Score"]=[92, 68, 87, 62, 90, 74]
df_subset["Average_Technical_Score"]=(
    df_subset["Python_Score"]
    + df_subset["SQL_Score"]
    + df_subset["Pandas_Score"]
    + df_subset["NumPy_Score"]
) / 4
conditions=[
    df_subset["Average_Technical_Score"]>=90,
    df_subset["Average_Technical_Score"]>=75,
    df_subset["Average_Technical_Score"]>=60
]
choices=[
    "Excellent",
    "Good",
    "Average"
]
df_subset["Performance"]=np.select(
    conditions,
    choices,
    default="Needs Improvement"
)
print(df_subset)
```

Q14

```python
newdepartment=df.groupby("Department").agg(
    Number_of_Students=("Student_ID", "count"),
    Average_Final_Score=("Final_Score", "mean"),
    Maximum_Final_Score=("Final_Score", "max"),
    Minimum_Final_Score=("Final_Score", "min")
)
newdepartment=newdepartment .sort_values(
    by="Average_Final_Score",
    ascending=False
)
print(newdepartment )
```
